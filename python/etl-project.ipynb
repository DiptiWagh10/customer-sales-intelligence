```python
# Recreate the derived delivery features

orders_clean["approval_hours"] = (
    orders_clean["order_approved_at"]
    - orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / 3600

orders_clean["delivery_days"] = (
    orders_clean["order_delivered_customer_date"]
    - orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)

orders_clean["estimated_delivery_days"] = (
    orders_clean["order_estimated_delivery_date"]
    - orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)

# Compare calendar dates for actual delivery delay
orders_clean["delivery_delay_days"] = (
    orders_clean["order_delivered_customer_date"].dt.normalize()
    - orders_clean["order_estimated_delivery_date"].dt.normalize()
).dt.days

# Late delivery flag
orders_clean["late_delivery"] = (
    orders_clean["delivery_delay_days"] > 0
).astype("Int64")

# Orders that have not been delivered should remain unknown
orders_clean.loc[
    orders_clean["order_delivered_customer_date"].isna(),
    "late_delivery"
] = pd.NA

# Time features
orders_clean["order_year"] = (
    orders_clean["order_purchase_timestamp"].dt.year
)

orders_clean["order_month"] = (
    orders_clean["order_purchase_timestamp"].dt.month
)

print("Features created successfully!")
print(
    orders_clean[
        [
            "approval_hours",
            "delivery_days",
            "estimated_delivery_days",
            "delivery_delay_days",
            "late_delivery",
            "order_year",
            "order_month"
        ]
    ].head()
)
```

```python
# ============================================================
# LEAKAGE-FREE TEMPORAL DATASET
# ============================================================

# 1. Define prediction cutoff
cutoff_date = pd.Timestamp("2018-07-01")

# 2. Split orders into historical and future periods
pre_cutoff_orders = order_analytics[
    order_analytics["order_purchase_timestamp"] <= cutoff_date
].copy()

post_cutoff_orders = order_analytics[
    order_analytics["order_purchase_timestamp"] > cutoff_date
].copy()

print("Pre-cutoff orders:", len(pre_cutoff_orders))
print("Post-cutoff orders:", len(post_cutoff_orders))


# ============================================================
# 3. Create customer features using ONLY pre-cutoff data
# ============================================================

pre_customer = (
    pre_cutoff_orders
    .groupby("customer_unique_id")
    .agg(
        total_orders=("order_id", "nunique"),
        total_spend=("total_payment_value", "sum"),
        avg_order_value=("total_payment_value", "mean"),
        total_items=("total_items", "sum"),
        avg_items_per_order=("total_items", "mean"),
        avg_review_score=("review_score", "mean"),
        avg_delivery_days=("delivery_days", "mean"),
        delivered_orders=("late_delivery", "count"),
        late_orders=("late_delivery", "sum"),
        last_purchase_date=("order_purchase_timestamp", "max")
    )
    .reset_index()
)


# ============================================================
# 4. Calculate recency as of cutoff
# ============================================================

pre_customer["recency_days"] = (
    cutoff_date - pre_customer["last_purchase_date"]
).dt.days


# ============================================================
# 5. Calculate late-delivery rate
# ============================================================

pre_customer["late_delivery_rate"] = (
    pre_customer["late_orders"]
    / pre_customer["delivered_orders"]
)


# ============================================================
# 6. Find customers who purchased AFTER cutoff
# ============================================================

future_orders = (
    post_cutoff_orders
    .groupby("customer_unique_id")["order_id"]
    .nunique()
    .reset_index(name="future_orders")
)


# ============================================================
# 7. Attach future behavior to historical customer data
# ============================================================

pre_customer = pre_customer.merge(
    future_orders,
    on="customer_unique_id",
    how="left"
)


# ============================================================
# 8. Create prediction target
# ============================================================

pre_customer["future_orders"] = (
    pre_customer["future_orders"].fillna(0)
)

pre_customer["repeat_after_cutoff"] = (
    pre_customer["future_orders"] > 0
).astype(int)


# ============================================================
# 9. Basic validation
# ============================================================

print("\nCustomer dataset shape:", pre_customer.shape)

print("\nTarget distribution:")
print(
    pre_customer["repeat_after_cutoff"].value_counts()
)

print("\nTarget percentage:")
print(
    pre_customer["repeat_after_cutoff"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nSample:")
print(
    pre_customer[
        [
            "customer_unique_id",
            "total_orders",
            "total_spend",
            "recency_days",
            "future_orders",
            "repeat_after_cutoff"
        ]
    ].head(10)
)
```
